```python
import matplotlib.pyplot as plt

df[13].value_counts().sort_index().plot(kind="bar")

plt.xlabel("Original target class")
plt.ylabel("Number of patients")
plt.title("Distribution of Heart Disease Diagnosis")
plt.show()
```

## Target Definition

The original `num` attribute contains five classes:

- 0: no significant heart disease
- 1–4: heart disease categories

For this project, the prediction task is defined as binary classification:

- 0 → No disease
- 1–4 → Disease

This allows the model to answer the question:
"Does this patient have heart disease?"
rather than predicting the original five-category diagnosis.
